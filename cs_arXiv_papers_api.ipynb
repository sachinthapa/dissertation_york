```python
# contenders = ['cloud','neural','hallucination','attention',  'embedding', 'inference', 'pipeline', 'prompt']
def get_neighbours(t_word):
    if t_word in model_2020_42.wv:
        print(f'{t_word}:2020\n',model_2020_42.wv.most_similar(t_word, topn=10),end='\n')
        print(model_2020_777.wv.most_similar(t_word, topn=10),end='\n')
        print(model_2020_123.wv.most_similar(t_word, topn=10))
        
        print(model_2020_42.wv.get_vecattr(t_word, 'count'),
        model_2020_777.wv.get_vecattr(t_word, 'count'),
        model_2020_123.wv.get_vecattr(t_word, 'count'),end='\n'*2)
    else:
        print(f"{t_word} ABSENT from vocab 2020")
    
    if t_word in model_2012_42.wv:
        print(f'{t_word}:2012\n',model_2012_42.wv.most_similar(t_word, topn=10),end='\n')
        print(model_2012_123.wv.most_similar(t_word, topn=10),end='\n')
        print(model_2012_777.wv.most_similar(t_word, topn=10))

        print(model_2012_42.wv.get_vecattr(t_word, 'count'),
        model_2012_123.wv.get_vecattr(t_word, 'count'),
        model_2012_777.wv.get_vecattr(t_word, 'count'),end='\n'*2)
    else:
        print(f"{t_word} ABSENT from vocab 2012")
    
    if t_word in model_1990_42.wv:
        print(f'{t_word}:1990\n',model_1990_42.wv.most_similar(t_word, topn=5),end='\n')
        print(model_1990_123.wv.most_similar(t_word, topn=10),end='\n')
        print(model_1990_777.wv.most_similar(t_word, topn=10))

        print(model_1990_42.wv.get_vecattr(t_word, 'count'),
        model_1990_123.wv.get_vecattr(t_word, 'count'),
        model_1990_777.wv.get_vecattr(t_word, 'count'),end='\n'*2)
    else:
        print(f"{t_word} ABSENT from vocab 1990")
```
