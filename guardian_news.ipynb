```python
#  ======================================= LOAD SAVED MODELS ============================================

# contenders = ['cloud','neural','hallucination','attention']
from gensim.models import Word2Vec

guardian_42 = Word2Vec.load("./guardian/guardian_42.bin")
guardian_777 = Word2Vec.load("./guardian/guardian_777.bin")
guardian_123 = Word2Vec.load("./guardian/guardian_123.bin")
print("models loaded")

#  ======================================= LOAD SAVED MODELS ============================================
```

```python
def get_neighbours(t_word):
    if t_word in guardian_42.wv:
        print(f'{t_word}\n',guardian_42.wv.most_similar(t_word, topn=10),end='\n')
        print(guardian_777.wv.most_similar(t_word, topn=10),end='\n')
        print(guardian_123.wv.most_similar(t_word, topn=10))
        
        print(guardian_42.wv.get_vecattr(t_word, 'count'),end='\n'*2)
    else:
        print(f"{t_word} ABSENT from vocab")
```

```python
get_neighbours('cloud')
```

```python
get_neighbours('stream')
```

```python
get_neighbours('port')
```

```python
wv = guardian_42.wv
sorted_vocab = sorted(wv.index_to_key, 
                      key=lambda word: wv.get_vecattr(word, "count"), 
                      reverse=True)

tmp = [f"{word}: {wv.get_vecattr(word, 'count')}" for word in sorted_vocab[:100]]
print(', '.join(tmp))
```
